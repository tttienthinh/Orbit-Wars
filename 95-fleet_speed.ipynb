```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
def speed(ships):
    max_speed = 6
    return 1.0 + (max_speed - 1.0) * (math.log(ships) / math.log(1000)) ** 1.5

def speed_np(ships):
    max_speed = 6
    return 1.0 + (max_speed - 1.0) * (np.log(ships) / np.log(1000)) ** 1.5

speed(100), speed_np(100)
```

# Plot

```python
x = np.arange(1, 1025)
y = speed_np(x)

x_base_2 = np.array([1, 2, 4, 8, 16, 32, 64, 128, 256, 512, 1024])
y_base_2 = speed_np(x_base_2)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='Fleet Speed', color='blue')
plt.scatter(x_base_2, y_base_2, color='red', label='Powers of 2', zorder=5)
plt.xlabel('Number of Ships')
plt.ylabel('Fleet Speed')

plt.grid()
```
